```python
import pandas as pd
data = {
    "name": ["A", "B", "C"],
    "age": [1, 2, 3]
}
df = pd.DataFrame(data, index= [1, 2, 3])
```

```python
df["job"] = ["cook", "N/A", "cashier"]
print(df)
```

```python
new_row = pd.DataFrame([{"name": "D", "age": 4, "job": "engineer"}],
                       index = [4])
df = pd.concat([df, new_row])
print(df)
```
